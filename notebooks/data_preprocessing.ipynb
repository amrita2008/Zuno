```python
import pandas as pd
import numpy as np
```

```python
songs = pd.read_csv("dataset (1).csv")

songs.head()
```

```python
songs = songs[
[
'track_name',
'artists',
'track_genre',
'popularity'
]
]
```

```python
songs.columns = [
'song',
'artist',
'genre',
'popularity'
]
```

```python
songs = songs.drop_duplicates(
subset=['song','artist']
)
```

```python
lyrics = pd.read_csv(
"spotify_millsongdata.csv"
)

lyrics.head()
```

```python
lyrics = lyrics[
[
'song',
'artist',
'text'
]
]
```

```python
lyrics.rename(
columns={
'text':'lyrics'
},
inplace=True
)
```

```python
lyrics['lyrics'] = (
lyrics['lyrics']
.astype(str)
.str.replace('\r',' ')
.str.replace('\n',' ')
)
```

```python
songs['song'] = (
songs['song']
.astype(str)
.str.lower()
)

songs['artist'] = (
songs['artist']
.astype(str)
.str.lower()
)

lyrics['song'] = (
lyrics['song']
.astype(str)
.str.lower()
)

lyrics['artist'] = (
lyrics['artist']
.astype(str)
.str.lower()
)
```

```python
zuno = pd.merge(
songs,
lyrics,
on=['song','artist'],
how='inner'
)
```

```python
print(zuno.shape)

zuno.head()
```

```python
zuno.to_csv(
"zuno_dataset.csv",
index=False
)
```
