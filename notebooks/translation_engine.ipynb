```python
def translate_lyrics(lyrics):

    lyrics = str(lyrics)

    lyrics = lyrics[:500]

    prompt = f"""
    Translate the following lyrics into English.

    Only return the translated lyrics.

    Lyrics:
    {lyrics}
    """

    response = model.generate_content(
        prompt
    )

    return response.text
```

```python
def translate_song(song_name):

    song_data = df[
        df['song']
        .str.lower()
        ==
        song_name.lower()
    ]

    if len(song_data) == 0:

        print("Song Not Found")

        return None

    lyrics = song_data[
        'lyrics'
    ].values[0]

    translation = translate_lyrics(
        lyrics
    )

    return translation
```
