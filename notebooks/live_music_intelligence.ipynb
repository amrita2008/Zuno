```python
from deep_translator import GoogleTranslator

def translate_lyrics(text):

    text = text[:3000]

    translated = GoogleTranslator(
        source='auto',
        target='en'
    ).translate(text)

    return translated
```

```python
translated = translate_lyrics(
    lyrics
)

print(
    translated
)
```

```python
with open(
    "translated_lyrics.txt",
    "w",
    encoding="utf-8"
) as f:

    f.write(
        translated
    )
```
